### A quoi me servira ce notebook ?

Ce notebook vous servira à la fois d'**outil d'analyse** et de **support pédagogique**, en raison de ses explication des outils donnés.

### Pourquoi l'avoir créé ?

Ce projet traite principalement la BRVM, un marché émergeant et donc peut étudié de manière quantitative. Ainsi ce projet à pour objectif de vous donner un réél outil quantitatif qui vous suivra lors de vos analyses. Pour être tout à fait franc ce notebook m'est avant tout bénéfique car il sera original pour mon dossier.

### Qu'en est il du ***edge*** ?

Pour être honnête la matrice de corrélation ne vous offfrera aucun *edge* (1). Il ne s'agit que d'un outil qui vous donnera une meilleur vision du marché et des relations entre chaques actifs
